# Practice on Titanic Dataset

## Step 1: Load Titanic dataset

```python
import pandas as pd
df = pd.read_csv("../../dataset/titanic/train.csv")
```

## Step 2: Select only numerical columns

```python
df[["Age", "Fare"]]
```

## Step 3: Apply StandardScaler

```python
from sklearn.preprocessing import StandardScaler

standard = StandardScaler()

df[["Age_standard", "Fare_standard"]] = standard.fit_transform(df[["Age", "Fare"]])
```

## Step 4: Apply MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler

minmax = MinMaxScaler()

df[["Age_minmax", "Fare_minmax"]] = minmax.fit_transform(df[["Age", "Fare"]])
```

## Step 5: Compare

```python
print(df[[
    "Age",
    "Age_standard",
    "Age_minmax",
    "Fare",
    "Fare_standard",
    "Fare_minmax"
]].head(10))
```
