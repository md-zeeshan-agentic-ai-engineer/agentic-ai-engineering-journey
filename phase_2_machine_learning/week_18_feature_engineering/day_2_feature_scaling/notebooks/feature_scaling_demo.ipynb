# Feature Scaling

## 1. StandardScaler

```python
from sklearn.preprocessing import StandardScaler
import pandas as pd

df = pd.DataFrame({
    "Age":[20,25,30,35,40]
})

scaler = StandardScaler()

df["Age_scaled"] = scaler.fit_transform(df[["Age"]])

print(df)
```

## 2. MinMaxScaler

```python
from sklearn.preprocessing import MinMaxScaler
import pandas as pd

df = pd.DataFrame({
    "Salary":[30000,50000,80000,100000,120000]
})

scaler = MinMaxScaler()

df["Salary_scaled"] = scaler.fit_transform(df[["Salary"]])

print(df)
```

## 3. RobustScaler

```python
from sklearn.preprocessing import RobustScaler
import pandas as pd

df = pd.DataFrame({
    "Income":[20, 25, 30, 35, 1000]
})

scaler = RobustScaler()

df["Income_scaled"]=scaler.fit_transform(df[["Income"]])

print(df)
```

## Observations

- StandardScaler transformed the data to have a mean close to 0 and a standard deviation close to 1.
- MinMaxScaler scaled all values between 0 and 1.
- RobustScaler handled the outlier effectively by using the median and IQR instead of the mean and standard deviation.
- RobustScaler is the preferred choice when the dataset contains significant outliers.
