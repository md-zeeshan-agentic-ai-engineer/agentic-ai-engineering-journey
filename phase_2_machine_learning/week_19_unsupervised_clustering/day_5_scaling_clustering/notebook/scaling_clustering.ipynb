# 9. Create the Dataset

Start with:

```python
import pandas as pd

data = {
    "Income": [20000, 25000, 30000, 100000, 120000, 130000, 400000, 450000, 500000],
    "Age": [22, 24, 26, 35, 37, 39, 58, 62, 65]
}

df = pd.DataFrame(data)

print(df)
```

# 10. Select Features

```python
X = df[["Income", "Age"]]
```

# 11. Experiment A - K-Means Without Scaling

```python
from sklearn.cluster import KMeans

kmeans_row = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df["Cluster_Row"] = kmeans_row.fit_predict(X)

print(df)
```

# 12. Experiment B — Scale the Features

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print(X_scaled)
```

# 13. K-Means After Scaling

Now create another model:

```python
kmeans_scaled = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df["Cluster_Scaled"] = kmeans_scaled.fit_predict(X_scaled)

print(df)
```

### Compare Results:
In this dataset, K-Means produced the same group assignments with and without feature scaling. This happens because the data contains three clearly separated groups, so scaling does not change the final grouping. However, the feature values themselves change significantly after scaling, and scaling remains important for K-Means when features have different ranges.

### One-line conclusion:

Same clusters ≠ scaling is unnecessary.

It means that for this particular dataset, the grouping is stable even after scaling.

# 26. Analyze the Cluster Centers

For the scaled model:

```python
print(kmeans_scaled.cluster_centers_)
```

These centers are expressed in scaled units.

#### To understand them in the original units, you can transform them back:

```python
centers_original = scaler.inverse_transform(
    kmeans_scaled.cluster_centers_
)

print(centers_original)
```

This is extremely useful.

#### You can then say something like:

Average Age ≈ 24

Average Income ≈ $25,000


Cluster 1:

Average Age ≈ 37

Average Income ≈ $120,000


Cluster 2:

Average Age ≈ 62
