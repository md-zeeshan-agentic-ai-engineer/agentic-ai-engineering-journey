# 24. Your Day 3 Experiment

We now move from theory to experimentation.

Use a synthetic dataset.

```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans
```

Create data:

```python
X, _ = make_blobs(
    n_samples=300,
    centers=4,
    cluster_std=1.2,
    random_state=42
)
```

Visualize it:

```python
plt.scatter(X[:, 0], X[:, 1])
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.title("Synthetic Dataset")
plt.show()
```

You should see several groups of points.

# 25. Experiment 1 — K = 2

```python
kmeans_2 = KMeans(
    n_clusters = 2,
    random_state = 42,
    n_init = "auto"
)

labels_2 = kmeans_2.fit_predict(X)
```

Visualize:

```python
plt.scatter(
    X[:, 0],
    X[:, 1],
    c=labels_2
)

plt.scatter(
    kmeans_2.cluster_centers_[:, 0],
    kmeans_2.cluster_centers_[:, 1],
    marker="X",
    s=200
)

plt.title("K-Means with K = 2")
plt.show()
```

Inspect:

```python
print("Inertia:", kmeans_2.inertia_)
```

# 26. Experiment 2 — K = 3

```python
kmeans_3 = KMeans(
    n_clusters=3,
    random_state=42,
    n_init="auto"
)

labels_3 = kmeans_3.fit_predict(X)

print("Inertia:", kmeans_3.inertia_)
```

# 27. Experiment 3 — K = 4

```python
kmeans_4 = KMeans(
    n_clusters=4,
    random_state=42,
    n_init="auto"
)

labels_4 = kmeans_4.fit_predict(X)

print("Inertia:", kmeans_4.inertia_)
```

# 28. Experiment 4 — K = 5

```python
kmeans_5 = KMeans(
    n_clusters=5,
    random_state=42,
    n_init="auto"
)

labels_5 = kmeans_5.fit_predict(X)

print("Inertia:", kmeans_5.inertia_)
```

# 29. Build an Inertia Table

Instead of manually comparing everything:

```python
results = []

for k in [2, 3, 4, 5]:

    model = KMeans(
        n_clusters = k,
        random_state = 42,
        n_init = "auto"
    )

    model.fit(X)

    results.append({
        "K": k,
        "Inertia": model.inertia_
    })

for result in results:
    print(result)
```

You might get something conceptually like:

K = 2 → high inertia

K = 3 → lower inertia

K = 4 → much lower inertia

K = 5 → slightly lower inertia

The exact values depend on the generated dataset.
