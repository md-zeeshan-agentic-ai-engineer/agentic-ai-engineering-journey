```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

n_estimators_values = [10, 50, 100, 200]

results = []

for n in n_estimators_values:
    model = RandomForestClassifier(
        n_estimators=n,
        max_depth=None,
        min_samples_split=2,
        min_samples_leaf=1,
        max_features="sqrt",
        random_state=42
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    results.append((n, accuracy))

for n, accuracy in results:
    print(f"n_estimators={n} → Accuracy={accuracy:.4f}")
```

- 10 trees achieved 96.67% accuracy on the current test set.
- 50, 100, and 200 trees achieved 90.00% accuracy.
- Increasing the number of trees did not improve test accuracy in this experiment.
- More trees generally increase computational cost.
- The effect of n_estimators should be evaluated on appropriate validation data rather than assuming that more trees always produce higher accuracy.

# 🔥 Now Experiment 3 — Feature Randomness

```python
max_features_values = ["sqrt", "log2"]

results_features = []

for feature_setting in max_features_values:
    model = RandomForestClassifier(
        n_estimators=100,
        max_depth=None,
        min_samples_split=2,
        min_samples_leaf=1,
        max_features=feature_setting,
        random_state=42
    )

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    accuracy = accuracy_score(y_test, y_pred)

    results_features.append((feature_setting, accuracy))

for feature_setting, accuracy in results_features:
    print(
        f"max_features={feature_setting} "
        f"→ Accuracy={accuracy:.4f}"
    )
```
