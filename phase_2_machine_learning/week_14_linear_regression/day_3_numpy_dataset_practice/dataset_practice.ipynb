```python
import numpy as np

X = np.array([1, 2, 3, 4, 5])

print(X)

print(type(X))


X = X.reshape(5, 1)

print(X)
print(X.shape)
```

```python
X = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

y = np.array([0, 1, 0, 1])

print("X:")
print(X)

print("\ny:")
print(y)

print("X Shape:", X.shape)
print("y Shape:", y.shape)

print(X[0])
print(X[0][1])
print(y[2])
```

```python
X = np.array([
    [180, 80, 40],
    [170, 90, 30],
    [160, 90, 50]
])

y = np.array([1, 0, 1])

print("X:")
print(X)

print("\ny:")
print(y)

print("\nX Shape:", X.shape)
print("y Shape:", y.shape)

print(X[1])
print(X[2][0])
```

```python
arr = np.array([10, 20, 30, 40, 50, 60])

print("Shape:", arr.shape)

arr = arr.reshape(2, 3)

print(arr)
print(arr.shape)

print(arr[0])
print(arr[:, 1])
```
