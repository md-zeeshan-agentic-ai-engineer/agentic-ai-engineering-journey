```python
def gini_impurity(y):
    classes = set(y)
    total = len(y)

    gini = 1

    for cls in classes:
        probability = y.count(cls) / total
        gini -= probability ** 2

    return gini
```

## 💻 Number 15 — Code Verification
We have:

### 1. First: Use the Gini function from Number 13
If you already wrote this function, reuse it:

```python
def gini_impurity(y):
    classes = set(y)
    total = len(y)

    gini = 1

    for cls in classes:
        probability = y.count(cls) / total
        gini -= probability ** 2

    return gini
```

#### 2. Create the Parent and Child Nodes

```python
parent = ["Yes"] * 6 + ["No"] * 4

left = ["Yes"] * 5 + ["No"] * 1

right = ["Yes"] * 1 + ["No"] * 3
```

#### 3. Calculate Each Gini

```python
parent_gini = gini_impurity(parent)
left_gini = gini_impurity(left)
right_gini = gini_impurity(right)

print("Parent Gini:", parent_gini)
print("Left Gini:", left_gini)
print("Right Gini:", right_gini)
```

#### 4. Calculate Weighted Gini

```python
total = len(parent)

left_weight = len(left) / total
right_weight = len(right) / total

weighted_gini = (
    left_weight * left_gini
    + right_weight * right_gini
)

print("Weighted Gini:", weighted_gini)
```

#### 5. Calculate Gini Gain

```python
gini_gain = parent_gini - weighted_gini

print("Gini Gain:", gini_gain)
```
