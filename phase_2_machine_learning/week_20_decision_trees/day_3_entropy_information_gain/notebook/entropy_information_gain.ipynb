After doing the mathematics manually, verify it with Python:

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder

X = [["Sunny"],
     ["Sunny"],
     ["Sunny"],
     ["Rain"],
     ["Rain"],
     ["Rain"]]

y = ["No", "No", "Yes", "Yes", "Yes", "No"]

encoder = LabelEncoder()

X_encoded = encoder.fit_transform([row[0] for row in X]).reshape(-1, 1)
y_encoded = encoder.fit_transform(y)

model = DecisionTreeClassifier(
    criterion="entropy",
    random_state=42
)
model.fit(X_encoded, y_encoded)

print("Tree trained successfully.")
```

## Important:
You are using:

```python
criterion="entropy"
```

This tells the decision tree to use entropy-based splitting.
